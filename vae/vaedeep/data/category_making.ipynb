```python
import pandas as pd
```

```python
def filtering_class(csv, category):

    df = pd.read_csv(csv)
    df_filter = df[df['category'] == category ] 
    df_filter.to_csv(f'vae_{category}_{csv}',index=False)
    return df_filter

df_oxt = filtering_class('oxt_1.0_2.csv',0)
df_ser = filtering_class('ser_0.5_2.csv',0)
```
