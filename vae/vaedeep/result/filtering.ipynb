```python
import pandas as pd
```

```python
def filtering_result(csv, threshold):

    df = pd.read_csv(csv)
    df_filter = df[df['average_probability'] > threshold][['sequence','average_probability']].sort_values(by='average_probability', ascending=False)
    df_filter.to_csv(f'filter_check_{csv}',index=False)
    return df_filter

df_oxt1 = filtering_result('oxt_1.csv',0.8)
df_ser1 = filtering_result('ser_1.csv',0.8)
df_oxt0 = filtering_result('oxt_0.csv',0.8)
df_ser0 = filtering_result('ser_0.csv',0.8)
```

```python
df_oxt1_10 = df_oxt1.head(10)
df_oxt1_10["sequence"] = df_oxt1_10["sequence"].apply(lambda x: f"CCCCCC{x}CCCCCC")
df_oxt1_10.to_csv("10_class1_oxt_arch.csv", index=False)
```

```python
df_ser1_10 = df_ser1.head(10)
df_ser1_10["sequence"] = df_ser1_10["sequence"].apply(lambda x: f"CCCCCC{x}CCCCCC")
df_ser1_10.to_csv("10_class1_ser_arch.csv", index=False)
```
