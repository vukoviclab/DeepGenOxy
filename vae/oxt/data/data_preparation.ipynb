```python
import pandas as pd
```

```python
def remove_c(path):
    df = pd.read_csv(path)
    if len(df['sequence'][2]) > 18:
        df['sequence'] = df['sequence'].apply(lambda x:x[6:-6])
    return df
```

```python
df = remove_c('oxt_select_r6.csv')
df.to_csv('oxt_sr.csv', index=False)
```
