## prepare the dataset for prediction which were genrated by the VAE

```python
import pandas as pd
```

```python
def csv_read(path):
    df = pd.read_csv(path, usecols=['sequence'])
    df.drop_duplicates(subset ='sequence',inplace = True)
    return df
```

```python
path_of_csv = [
    'results/all_sequences_1.csv',
    'data/oxt.csv',
    'data/selec.csv',
    'data/selec_filtered.csv',
]
```

```python
frames = []
for path in path_of_csv:
    frames.append(csv_read(path))
```

```python
df_new = frames[0][~frames[0]['sequence'].isin(frames[1]['sequence'])]
df_new = df_new[~df_new['sequence'].isin(frames[2]['sequence'])]
df_new = df_new[~df_new['sequence'].isin(frames[3]['sequence'])]
df_new.to_csv('data/vae_prediction.csv', index = False)
```
