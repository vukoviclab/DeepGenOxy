```python
import pandas as pd

def find_most_overlapping_file(main_file, file1, file2):
    """
    Determines which of two CSV files has more overlapping sequences with a main CSV file.

    Args:
        main_file (str): The path to the main CSV file.
        file1 (str): The path to the first CSV file to compare.
        file2 (str): The path to the second CSV file to compare.

    Returns:
        str: A message indicating which file has more overlap.
    """
    try:
        # Load the datasets
        main_df = pd.read_csv(main_file)
        df1 = pd.read_csv(file1)
        df2 = pd.read_csv(file2)

        # Extract the 'sequence' column from each DataFrame into a set for efficient comparison
        main_sequences = set(main_df['sequence'])
        sequences1 = set(df1['sequence'])
        sequences2 = set(df2['sequence'])

        # Find the intersection (common sequences)
        common_with_file1 = main_sequences.intersection(sequences1)
        common_with_file2 = main_sequences.intersection(sequences2)

        # Get the number of overlapping sequences
        overlap_with_file1 = len(common_with_file1)
        overlap_with_file2 = len(common_with_file2)

        # Print the overlap counts
        print(f"Number of overlapping sequences with {file1}: {overlap_with_file1}")
        print(f"Number of overlapping sequences with {file2}: {overlap_with_file2}")

        # Determine and announce which file has more overlap
        if overlap_with_file1 > overlap_with_file2:
            return f"{file1} has more overlap with {main_file}."
        elif overlap_with_file2 > overlap_with_file1:
            return f"{file2} has more overlap with {main_file}."
        else:
            return "Both files have the same amount of overlap."

    except FileNotFoundError as e:
        return f"Error: {e}. Please make sure all CSV files are in the correct directory."
    except KeyError as e:
        return f"Error: {e}. Make sure the 'sequence' column exists in all CSV files."

# --- Main execution ---
if __name__ == "__main__":
    main_csv = 'result_nn_vae.csv'
    csv_file1 = 'all_sequences.csv'
    csv_file2 = 'all_sequences_1.csv'

    result = find_most_overlapping_file(main_csv, csv_file1, csv_file2)
    print(result)
```
