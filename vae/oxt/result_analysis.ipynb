```python
import pandas as pd
```

```python
df_vae = pd.read_csv('results/result_nn_vae.csv')
df_vae.sort_values(by='average_probability', ascending=False, inplace=True)
df_vae_result = df_vae[['sequence','average_probability','max_probability','min_probability','architecture']]
df_vae_result_75 =  df_vae_result[df_vae_result['average_probability']> 0.80]
# df_vae_result_75.to_csv('result_vae_80.csv', index = False)
df_vae_result_75
```

```python
top2 = df_vae_result_75.groupby("architecture").apply(lambda x: x.nlargest(2, "average_probability")).reset_index(drop=True)
top2
```

```python
df_r6 = pd.read_csv('results/result_nn_r6.csv')
df_r6.sort_values(by='average_probability', ascending=False, inplace=True)
df_r6_result = df_r6[['sequence','average_probability','max_probability','min_probability','reads','row_number']]
df_r6_result_75 =  df_r6_result[df_r6_result['average_probability']> 0.80]
df_r6_result_75.to_csv('result_r6_80.csv', index = False)
df_r6_result_75
```
