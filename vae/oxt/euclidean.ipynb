```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('plotty.csv')
df
```

```python
target_row = df.iloc[104, :]
df['dist'] = (df[['PCA 1', 'PCA 2']] - target_row[['PCA 1', 'PCA 2']]).pow(2).sum(axis=1).pow(0.5)
sorted_df = df.sort_values(by='dist')
sorted_df
```

```python
df_vae_80 = pd.read_csv('result_vae_80.csv')
merged_df = pd.merge(sorted_df, df_vae_80, on='sequence', how='inner')
merged_df
```

```python
merged_df.to_csv('distance.csv',index=False)
```

```python
df_send = pd.read_csv('vae.csv')
df_send
```

```python
df_send_all_info = pd.merge(df_send,sorted_df, on='sequence', how='inner')
df_send_all_info
```

```python
df_send_all_info_1 = pd.merge(df_send_all_info,df_vae_80, on='sequence', how='inner')
df_send_all_info_1
```

```python
df_send_all_info_final = df_send_all_info_1[['sequence','dist','average_probability','max_probability','min_probability','architecture']]
df_send_all_info_final
```

```python
df_send_all_info_final.to_csv('result_vae_send.csv',index=False)
```
