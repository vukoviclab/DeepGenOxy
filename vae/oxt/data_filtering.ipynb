```python
import pandas as pd
```

```python
def exist_checking(df, df_main, column_name):
    df['temp'] = 1
    df_main = df_main.merge(df[['sequence', 'temp']], on='sequence', how='left')
    df_main['temp'].fillna(0, inplace=True)
    df_main.rename(columns={'temp': column_name}, inplace=True)
    return df_main
```

```python
df_result_vae = pd.read_csv('results/result_vae_all.csv')
df_result_r6 = pd.read_csv('results/result_nn_all.csv')
```

```python
df_data_vae = pd.read_csv('data/all_sequences_1.csv')
df_data_r6 = pd.read_csv('data/select_r6_reads.csv')
```

```python
df_result_vae = df_result_vae.merge(df_data_vae[['sequence', 'architecture']], on='sequence', how='left')
df_result_vae
```

```python
df_data_r6['row_number'] = df_data_r6.reset_index().index + 1
df_result_r6 = df_result_r6.merge(df_data_r6[['sequence', 'reads','row_number']], on='sequence', how='left')
df_result_r6
```

```python
df_result = pd.concat([df_result_r6, df_result_vae], ignore_index=True, sort=False)
df_result.fillna(-1, inplace=True)
df_result
```

```python
df_c6 = pd.read_csv('data/selec_c6.csv')
df_r6 = pd.read_csv('data/selec_r6.csv')
df_dff = pd.read_csv('data/oxt.csv')
df_vae = pd.read_csv('data/vae_prediction.csv')
dict_frames = {'control':df_c6,'selec_R6':df_data_r6, 'DFF':df_dff, 'VAE':df_vae}
```

```python
df_result
```

```python
for key, value in dict_frames.items():
    value['temp'] = 1
    df_result = df_result.merge(value[['sequence', 'temp']], on='sequence', how='left')
    df_result['temp'].fillna(0, inplace=True)
    df_result.rename(columns={'temp': key}, inplace=True)
    
```

```python
df_result.to_csv('final_result.csv', index=False)
```
